```python
%matplotlib inline

import os
os.environ["KMP_DUPLICATE_LIB_OK"]="TRUE" # tells OpenMP to not complain if it notices that two copies of OpenMP are loaded.

import matplotlib.pyplot as plt
```

```python
from pathlib import Path

from ultralytics import FastSAM
from ultralytics import settings
print(settings)
```

```python
data_dirpath=Path("D:/results/sam")
settings['datasets_dir']=str(data_dirpath)
settings['weights_dir']=str(data_dirpath/'fastsam'/'checkpoints')
settings['runs_dir']=str(data_dirpath/'fastsam')
# settings['weights_dir'].mkdir(parents=True, exist_ok=True)
if not os.path.isdir(settings['weights_dir']): os.makedirs(settings['weights_dir'])

source=data_dirpath/'bus.jpg'

# create a model
model=FastSAM('FastSAM-s.pt')

# run inference on an image
everything_results=model(source, device='cpu', retina_masks=True, imgsz=1024, conf=0.4, iou=0.9, 
                         save=True,   # You must set save=True to generate a folder
                         exist_ok=True)   # Set to True to prevent 'mask_test2', 'mask_test3', etc.)
```

```python
# run inference with points prompt
results=model(source, points=[[481,573]], labels=[1], save=True, exist_ok=True)
```

```python
# run inference with bboxes prompt
results=model(source, bboxes=[[439, 437, 524, 709]], save=True, exist_ok=True)
```

```python
# run inference with texts prompt
results=model(source, texts='bus', save=True, exist_ok=True)
```

```python
# run inference with bboxes, points, and texts
results=model(source, bboxes=[[439, 437, 524, 709]], points=[[481,573]], labels=[1], texts='bus', save=True, exist_ok=True)
```
