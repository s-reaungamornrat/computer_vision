```python
from pathlib import Path

import requests
from IPython.display import Audio

def play_audio(samples):
    return Audio(samples.data, rate=samples.sample_rate)
```

```python
source='file'
raw_audio_bytes=None

if source=='file':
    data_dirpath=Path('D:/data/kinetics')
    raw_audio_bytes=data_dirpath/'_WWyCxEok2I_000078_000088.mp4'
else:
    # Audio source is CC0: https://opengameart.org/content/town-theme-rpg
    # Attribution: cynicmusic.com pixelsphere.org
    url = "https://opengameart.org/sites/default/files/TownTheme.mp3"
    response=requests.get(url, headers={"User-Agent":""})
    if response.status_code!=200: raise RuntimeError(f'Failed to download vide. {response.status_code=}.')
    raw_audio_bytes=response.content
```

```python
# Decoder input can be (1) raw (encoded) audio bytes; (2) local audio file path; (3) audio stream from videos

from torchcodec.decoders import AudioDecoder

decoder=AudioDecoder(raw_audio_bytes)
print(decoder.metadata)
```

```python
samples=decoder.get_all_samples()
print(samples)
play_audio(samples)
```

```python
# .data is tensor of shape (num_channels, num_samples) of float dtype with values in [-1, 1] 
print('samples.data  (n_channels, n_samples) shape ', samples.data.shape,  ' type ', type(samples.data), samples.data.dtype, 
      ' range ', samples.data.min(), samples.data.max())
```

```python
samples=decoder.get_samples_played_in_range(start_seconds=0.07, stop_seconds=6)
print(samples)
play_audio(samples)
```

```python
decoder=AudioDecoder(raw_audio_bytes, sample_rate=16_000)
samples=decoder.get_all_samples()

print(samples)
play_audio(samples)
```
