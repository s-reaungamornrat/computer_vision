```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt

#from torchvision.utils import make_grid
```

```python
# LinearParamScheduler
end_factor=1.
start_factor=1./3.
total_iters=250
steps=np.arange(-1, total_iters)
y=(1.+(end_factor-start_factor)/(total_iters*start_factor + (steps-1)*(end_factor-start_factor)) )
plt.plot(steps, y)
```
