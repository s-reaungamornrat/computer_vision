```python
import os
from pathlib import Path
```

```python
data_dirpath=Path('D:/data/UCF101')
video_dirpath=data_dirpath/'UCF-101'
annotation_dirpath=data_dirpath/'UCF101TrainTestSplits-RecognitionTask'
class_id_fpath=annotation_dirpath/'classInd.txt'
test_fpath=annotation_dirpath/'testlist01.txt'
train_fpath=annotation_dirpath/'trainlist01.txt'
```

```python
with open(class_id_fpath, 'r') as f: class_id_info = f.readlines()
class_ids=dict([int(item) if i==1 else item for i, item in enumerate(line.strip().split()[::-1])] for line in class_id_info)
```

```python
with open(train_fpath, 'r') as f: content = f.readlines()
for line in content:
    file, label=line.strip().split()
    class_name=os.path.dirname(file)
    class_id=int(label)
    assert class_id==class_ids[class_name], f"Class name {class_name} is with provided ID {class_ids[class_name]} but got {class_id}"
```

```python
with open(test_fpath, 'r') as f: content = f.readlines()
outputs=[]
for line in content:
    line=line.strip()
    class_name=os.path.dirname(line)
    outputs.append(f'{line} {class_ids[class_name]}')
outputs='\n'.join(outputs)
with open(annotation_dirpath/'vallist01.txt', 'w') as f: f.write(outputs)
```
