```python
import numpy as np
```

```python
thresh=128
add=10

lut_ = []
for i in range(256):
    if i < thresh:
        lut_.append(min(255, i + add))
    else:
        lut_.append(i)
print(lut_)
```

```python
lut=np.arange(0, 256)
focus=lut<thresh
lut[focus]=np.minimum(255,lut[focus]+add)
np.allclose(lut_, lut)
```
